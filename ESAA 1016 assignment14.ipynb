# Chapter 05. 합성곱 신경망
01. 합성곱 신경망

합성곱 신경망: 이미지 전체를 한 번에 계산하는 것이 아닌 이미지의 국소적 부분을 계산함으로써 시간과 자원을 절약하여 이미지의 세밀한 부분까지 분석할 수 있는 신경망

### 5.1.1 합성곱층의 필요성

이미지, 영상 처리에 유용

배열을 펼쳐서
각 픽셀에 가중치 곱
은닉층으로 전달.
단 데이터의 공간적 구조를 무시하게 됨 -> 합성곱층 필요

### 5.1.2 합성곱 신경망 구조

음성 인식이나 이미지/영상 인식에서 주로 사용되는 신경망


다차원 배열 데이터를 처리하도록 구성되어 컬러 이미지 같은 다차원 배열 처리에 특화되어 있으며, 다섯 개의 계층으로 구성됨


입력층 / 합성곱층 / 풀링층 / 완전연결층 / 출력층



**입력층**

이미지: 높이, 너비, 채널 값 갖는 3차원 데이터

채널: gray scale이면 1, RGB면 3 값 가짐.


**합성곱층**

커널/필터 이용해 입력 데이터에서 특성 추출
-> '특성 맵' 추출, 크기가 줄어듦

- gray scale 이미지
일반적으로 커널은 3x3, 5x5 크기.
스트라이드(stride)라는 지정된 간격에 따라 순차적 이동.
입력 이미지&필터 포개 대응되는 숫자끼리 곱한 후 모두 더함

- RGB 이미지
필터 채널이 3(단, 필터 개수 3개 아니고 1개!!)
RGB 각각에 서로 다른 가중치로 합성곱 적용 후 결과 더해줌.

필터가 2개 이상이면
필터 각각이 특성 추출 결과의 채널이 됨.

**풀링층**
- 2,3번 반복으로 주요 특성 벡터 추출

특성 맵의 차원을 다운 샘플링하여 연산량 감소

주요 특성 벡터 추출(효과적 학습)

**완전연결층 **
- 1차원 벡터로 변환

출력층 **굵은 텍스트**
- 활성화함수 사용해 최종 결과 출력

### 5-1-3. 1D, 2D, 3D 합성곱
이동하는 방향의 수와 출력 형태에 따라 1D, 2D, 3D로 분류

**1D 합성곱**
- 필터가 시간을 축으로 좌우로만 이동할 수 있는 합성곱. 입력(W)과 필터(k)에 대한 출력은 W

**2D 합성곱**
- 필터가 방향 두 개로 움직이는 형태. 즉, 입력(W, H)과 필터(k, k)에 대한 출력은 (W, H)가 되며, 출력 형태는 2D 행렬

**3D 합성곱 **
- 필터가 움직이는 방향이 세 개. 입력(W, H, L)에 대해 필터(k, k, d)를 적용하면 출력으로 (W, H, L)을 가짐. d < L을 유지하는 것이 중요

**3D 입력을 갖는 2D 합성곱 **
- 필터 길이 = 입력 채널 길이일 때,
두 방향으로 움직임

1 * 1 합성곱 **굵은 텍스트**
-  3D 형태 입력(W, H, L) + 필터(1,1,L) -> 출력(W,H) 1*1 합성곱에서 채널 수를 조정해서 연산량이 감소되는 효과가 있으며, 대표적 사례로는 GoogleNet이 있음
